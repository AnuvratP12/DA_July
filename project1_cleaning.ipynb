```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset('titanic')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns = 'deck')
```

```python
df.isnull().sum()
```

```python
df['age'].median()
```

```python
df['age']=df['age'].fillna(df['age'].median())
```

```python
df.isnull().sum()
```

```python
df['embarked'].mode()[0]
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df.isnull().sum()
```

Data cleaning.

Explored the titanic dataset and did some basic cleaning on it. Tackled the missing values problem seen in columns like age, deck and embarked. All these problems were solved using different approaches. For age , I used median to fill the missing values. For deck , I dropped the columns because the percentage of missing data was massive and for embarked , I used mode to fill the missing values as the percentage was quite low in this.
